**¿Por qué es arquitectónicamente seguro eliminar las filas donde falta Android Ver, pero sería un error fatal eliminar las filas donde falta Rating?**

```python
#rellenar los NaN de la columna Rating usando la mediana de su propia categoría.
df['Rating'] = df['Rating'].fillna(df.groupby('Category')['Rating'].transform('median'))
```

```python
#replace text with NaN

df['Size'] = df['Size'].replace('Varies with device', np.nan)
```

**¿Por qué es mejor convertirlo a NaN en esta etapa en lugar de ponerle un tamaño falso como "0 MB"?**

Es mejor convertirlo a Nan porque si se remplazan con 0, esto introduce sesgo a la izquierda, mientras que NaN va a ser excluido en las operaciones.
