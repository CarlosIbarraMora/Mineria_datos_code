```python
import pandas as pd
import numpy as np

url = "https://raw.githubusercontent.com/mricardo89/data-mining/main/Unit-II/datasets/googleplaystore.csv"

df = pd.read_csv(url)
```

```python
df.isnull().sum()

```

```python
df.count()
```

```python
df.dropna(subset=['Android Ver', 'Type'])
```

```python
df.count()
```

```python
#rellenar los NaN de la columna Rating usando la mediana de su propia categoría.
df['Rating'] = df['Rating'].fillna(df.groupby('Category')['Rating'].transform('median'))
```
