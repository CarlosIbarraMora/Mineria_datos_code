```python
import pandas as pd
import numpy as np
url = "https://raw.githubusercontent.com/mricardo89/data-mining/main/Unit-II/datasets/googleplaystore.csv"
df = pd.read_csv(url)
```

```python
df.duplicated(subset=['App']).sum()
```

```python
df.drop_duplicates(subset=['App'], keep='first', inplace=True)
```

Salieron 1181 registros duplicados

```python
df['Installs'] = (
    df['Installs']
    .str.replace('+', '', regex=False)
    .str.replace(',', '', regex=False)
)

#Tuve que agregar esta linea debido a que esa columna contiene palabras y no solo números
# Por lo que esto hace que se intente convertir a numerico y si no puede lo convierte en NaN
df['Installs'] = pd.to_numeric(df['Installs'], errors='coerce')

df['Installs'].mean()
```

```python
df['Price'] = df['Price'].str.replace('$', '')


df['Price'] = pd.to_numeric(df['Price'], errors='coerce').astype(float)
```

```python
mask = df[df['Price'] > 200]

print(mask)
```

```python
clean = df[df['Price'] < 50]

print(clean)
```
