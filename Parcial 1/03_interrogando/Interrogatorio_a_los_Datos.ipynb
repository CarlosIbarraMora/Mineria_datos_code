```python
import pandas as pd
 
 
url = "https://raw.githubusercontent.com/mricardo89/data-mining/main/Unit-II/datasets/titanic.csv"
df = pd.read_csv(url)
```

```python
df["Survived"].value_counts(normalize=True)
```

```python
df.groupby("Sex")["Survived"].value_counts(normalize=True)
```

```python
import numpy as np

Q1, Q3 = np.percentile(df["Fare"].dropna(), [25, 75])

print("Q1", Q1)
print("Q3", Q3 )
```

```python
IQR = Q3 - Q1

print(IQR)
```

```python
sup_lim = Q3 + 1.5 * IQR
print(sup_lim)
```

```python
df[df["Fare"] > sup_lim].groupby("Pclass")["Fare"].count()
```

```python
mean = np.mean(df["Fare"])
median = np.median(df["Fare"])

print("Mean:", mean)
print("Median: ", median)
```

```python

# Proporciones originales 61% = 0, 39% = 1
proportions = df["Survived"].value_counts(normalize=True)

n_0 = round(proportions[0] * 150)
n_1 = 150 - n_0

sample_0 = df[df["Survived"] == 0].sample(n=n_0, random_state=42)
sample_1 = df[df["Survived"] == 1].sample(n=n_1, random_state=42)

sample = pd.concat([sample_0, sample_1])

print(sample["Survived"].value_counts())
print(sample["Survived"].value_counts(normalize=True))
```
