```python
import csv

totalF = 0
entryF = 0
totalM = 0
entryM = 0

with open('../datasets/titanic.csv', mode='r') as file:
    reader = csv.DictReader(file)

    for row in reader:
        if row["Sex"] == "female":
            entryF += 1
            totalF += float(row["Fare"])
        else:
            entryM += 1
            totalM += float(row["Fare"])

averageF = totalF / entryF
averageM = totalM / entryM


print("Female average: ", averageF)
print("Male average: ", averageM)
```

```python
import pandas as pd

data = pd.read_csv("../datasets/titanic.csv")

data.groupby("Sex")["Fare"].mean()
```

```python
import numpy as np

left_outliers = 0
right_outliers = 0

q1, q3 = np.percentile(data["Fare"].dropna(), [25, 75])

inter = q3 - q1

minimum = q1 - 1.5 * inter
maximum = q3 + 1.5 * inter

for fare in data["Fare"]:
    if fare < minimum:
        left_outliers += 1

    if fare > maximum:
        right_outliers += 1

print("Left outliers:", left_outliers)
print("Right outliers:", right_outliers)
```
