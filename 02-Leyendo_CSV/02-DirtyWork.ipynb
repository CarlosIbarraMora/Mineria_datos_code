```python
import random

random_list = [random.random() for _ in range(50_000_000)]

print("Random list length: ", len(random_list))
```

```python
def std_deviation():
    total = 0
    N = len(random_list)
    for i in random_list:
        total += i

    mean = total / N

    sum = 0
    for i in range(0, N):
        sum += ((random_list[i] - mean)**2)

    SD = (sum / N)**0.5

    print('std: ', SD)
std_deviation()
```

```python
import numpy as np

arr = np.array(random_list)


print(f'std with numpy: {arr.std()}')
```
